```python
# Binance API endpoint for ping (used to test connectivity)
url = "https://api.binance.com/api/v3/ping"

# Make the request
response = requests.get(url)

# Check response headers for rate limit information
retry_after = response.headers.get('Retry-After', 'Not Applicable')
used_weight = response.headers.get('X-MBX-USED-WEIGHT-1m', 'Unknown')
used_weight_1d = response.headers.get('X-MBX-USED-WEIGHT-1d', 'Unknown')

print(f"Current used weight (1 minute interval): {used_weight}")
print(f"Current used weight (1 day interval): {used_weight_1d}")
# Additional headers related to rate limits can also be inspected:
# X-MBX-ORDER-COUNT-<intervalNum><intervalLetter>
# X-MBX-USED-WEIGHT-<intervalNum><intervalLetter>
```

```python
# Fetch a certain ticker on the Binance API
def get_ticker_data(symbol, interval):
    base_url = "https://api.binance.com"
    endpoint = "/api/v3/klines"
    params = {'symbol': symbol, 'interval': interval}
    url = f"{base_url}{endpoint}"
    response = requests.get(url, params=params)
    data = response.json()
    # Pretty print the JSON data
    print(json.dumps(data, indent=4, sort_keys=True))
    # Access and print the X-MBX-USED-WEIGHT-1m header
    used_weight = response.headers.get('X-MBX-USED-WEIGHT-1m', 'No weight data available')
    print(f"Current used weight (in the past minute): {used_weight}")

    return data

# Example usage:
symbol = "BTCUSDT"  # Replace with your symbol
interval = "1d"
ticker_data = get_ticker_data(symbol,interval)
#print(ticker_data) # Print the Python fetched dictionary
```
